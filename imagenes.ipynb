1. importar librerias

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from  google.colab import files
from PIL import Image, ImageOps
```

2. cargar el conjunto de datos


```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()


x_train, x_text =  x_train / 255.0, x_test / 255.0


print(f"datos de entrenamiento:{x_train.shape[0]}")
print(f"datos de prueba:{x_test.shape[0]}")
```

3. crear modelo y entrenarlo

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax'),
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print('modelo creado exitosamente')
model.summary()
```

4. entrenar el modelo


```python
from os import access
print("entrenando el modelo...")
historial = model.fit(x_train, y_train, epochs=5,)
loss, accuracy = model.evaluate(x_text, y_test)
print(f"perdida:{loss}")
print(f"exactitud:{accuracy*100:.2f}%")
```

5. probar el modelo

```python
print("suba una imagen:")
uploaded = files.upload()

for filename in uploaded.keys():
  img = Image.open(filename).convert("L")
  img_invertida = ImageOps.invert(img)
  img_redimensionada = img_invertida.resize((28, 28))
  img_array = np.array(img_redimensionada)

  img_array = np.array(img_redimensionada) / 255.0
  img_lista_para_predecir =np.expand_dims(img_array, axis=0)

  predicciones = model.predict(img_lista_para_predecir)
  numero_predicho = np.argmax(predicciones)
  confianza = np.max(predicciones)

  print(f"el numero espredicho: {numero_predicho}")
  print(f"con una confianza del: {confianza*100:.2f}%")
```
